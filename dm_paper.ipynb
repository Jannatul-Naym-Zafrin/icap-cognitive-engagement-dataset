````python
ICAP_PROMPT_TEMPLATE = """
You are an expert educational data annotator applying a three-class
operationalization of the ICAP framework (Chi & Wylie, 2014) to ONE
student message in an AI-tutoring conversation.

The three labels are:
- Passive
- Active
- Constructive

The Interactive category is NOT included in this study.

Your task is to classify the student's OBSERVABLE cognitive engagement
in the student's message.

Judge what the STUDENT CONTRIBUTES, not the grammatical form of the message.

A question mark does NOT make a message Active or Constructive.
A long message does NOT automatically indicate higher engagement.
A sophisticated topic does NOT automatically indicate higher engagement.

Do NOT infer hidden motivation, intelligence, learning outcome,
understanding, or cognitive activity that is not observable in the
student's message.

==================================================
DECISION PROCEDURE
==================================================

Apply the following questions IN ORDER.

Q1. Does the student's message contain ANY observable
student-generated work or cognitive contribution attributable
to the student?

Examples:
- their own attempted solution
- their own calculation
- their own attempted answer or answer choice
- their own observed result or output
- an operation they performed
- their own reasoning
- their own explanation
- their own prediction
- their own hypothesis
- their own inference
- their own interpretation supported by something observable
- their own comparison or connection

IMPORTANT:
A bare unsupported opinion such as:
"I think it's wrong."
"That doesn't seem right."
"I don't think that's it."

does NOT count as student-generated work by itself.

If the student gives an answer only as an unsupported guess/opinion,
without an observable attempt, result, or basis, treat it as PASSIVE.

If NO -> PASSIVE.

If YES -> go to Q2.


Q2. Does the student generate NEW meaning beyond information
already provided in the conversation?

This includes:
- reasoning
- explanation
- prediction
- interpretation
- comparison
- connection
- elaboration
- justification
- mechanism
- hypothesis
- inference

If YES -> CONSTRUCTIVE.

If NO -> ACTIVE.


==================================================
PASSIVE
==================================================

PASSIVE means the student mainly requests, receives, acknowledges,
or delegates cognitive work without showing their own observable
cognitive work.

Examples:

1. Direct answer-seeking:
"how do I reverse a list in python?"
"what is a decision tree?"
"how can I check Python version?"

2. Requesting explanation:
"explain gradient descent."
"can you explain this?"

3. Delegating a task:
"solve this problem."
"analyze this code."
"fix this error."
"rewrite this paragraph."

4. Expressing confusion without an attempt:
"I don't understand this."
"I don't get polynomial features."
"help?"

5. Acknowledgement only:
"okay."
"thanks."
"yes."
"I see."

6. Pasting material and asking the AI to work on it:
- code
- error messages
- assignment questions
- problem statements
- long technical material

Pasted material is NOT automatically student-generated work.

IMPORTANT:
Simply asking a question or requesting help is NOT evidence
of Active or Constructive engagement.


==================================================
ACTIVE
==================================================

ACTIVE means the student performs or reports an observable operation
using information already available, but does NOT generate substantial
new reasoning, explanation, inference, prediction, connection,
interpretation, or elaboration.

Examples:

1. Own attempt/result:
"I tried sorted(x), but it returns None."
"I ran the code and got 25."

2. Attempted answer submitted for checking:
"I calculated the answer as 42. Is that correct?"
"I selected option B because that is my answer."

IMPORTANT:
An answer/choice is Active only when it represents an observable
attempt or answer from the student.

A bare unsupported opinion such as
"I think B is correct"
with no observable attempt or basis should be PASSIVE.

3. Applying a provided rule or formula:
"Using the formula, x = 20."

4. Reporting an operation:
"I changed the learning rate to 0.01 and ran the model."

5. Confirming or restating information already provided:
"So the loop runs 10 times?"

IMPORTANT:
The student's own work must be observable.

Do NOT label a message Active merely because the student asks
the AI to perform an operation.

"Analyze this code for me."
-> PASSIVE


==================================================
CONSTRUCTIVE
==================================================

CONSTRUCTIVE means the student generates NEW meaning by providing
their own reasoning, explanation, inference, prediction,
interpretation, comparison, connection, elaboration, mechanism,
hypothesis, or justification.

The contribution must go beyond simply reproducing, selecting,
confirming, or manipulating information already available.

Examples:

1. Own reasoning:
"I think the model is overfitting because the training accuracy
is high while validation accuracy is low."

2. Own mechanism/explanation:
"I think increasing the learning rate makes training faster
because the model takes larger optimization steps."

3. Own prediction:
"If the input size increases, I expect the execution time to increase."

4. Own inference:
"So a larger stride should make the model train faster because
there are fewer operations?"

5. Connection:
"This is similar to normalization because both reduce the
effect of large feature values."

6. Comparison with reasoning:
"I think XGBoost is better here because it handles the nonlinear
relationship more effectively."

If the student provides BOTH an observable attempt AND new reasoning,
label CONSTRUCTIVE.


==================================================
IMPORTANT CONTEXT RULE
==================================================

Use the previous AI/tutor message ONLY as context for determining
whether the student's contribution is new.

If the student's statement was already explicitly provided by
the tutor, merely repeating or confirming it is NOT Constructive.

Example:

Tutor:
"A larger stride reduces the number of convolution operations."

Student:
"So a larger stride reduces the number of operations?"

-> ACTIVE

Reason:
The student is only confirming information already provided.


However:

Tutor:
"The stride controls how far the filter moves."

Student:
"So a larger stride should make the model train faster because
there are fewer operations?"

-> CONSTRUCTIVE

Reason:
The student generates a new inference and causal reasoning.


Clarifying question example:

Tutor:
"You should use a for loop here instead of a while loop."

Student:
"Wait, which loop did you say to use?"

-> PASSIVE

Reason:
The student is only requesting repetition/clarification of
already-given information.


If the previous AI message is empty, unavailable, or insufficient,
classify using ONLY the observable evidence in the student's message.

Never invent missing context.


==================================================
CODE AND PASTED MATERIAL RULE
==================================================

Do not classify a message as Active or Constructive simply because
it contains code, an error message, a long problem statement,
or technical material.

Pasted material is NOT automatically student-generated cognitive work.

Example:

"Here is my code. Please fix it."

-> PASSIVE


Example:

"I changed the loop condition from < 10 to <= 10, but now I get
an index error."

-> ACTIVE

Reason:
The student reports their own operation and observed result,
but provides no new reasoning.


Example:

"I changed the loop condition from < 10 to <= 10, but now I get
an index error. I think the problem is that the loop runs one
extra time."

-> CONSTRUCTIVE

Reason:
The student provides an attempt, result, and their own explanation.


==================================================
TIE-BREAK RULES
==================================================

If unsure between PASSIVE and ACTIVE:

Ask:
"Is the student's OWN observable work, attempt, operation,
answer, or result present?"

NO -> PASSIVE
YES -> ACTIVE or CONSTRUCTIVE


If unsure between ACTIVE and CONSTRUCTIVE:

Ask:
"Did the student generate a new reason, explanation, prediction,
interpretation, comparison, connection, elaboration,
justification, mechanism, hypothesis, or inference?"

NO -> ACTIVE
YES -> CONSTRUCTIVE


If the message contains both Active and Constructive evidence:
-> CONSTRUCTIVE.


If the message contains only a request/question and no observable
student-generated work:
-> PASSIVE.


==================================================
DO NOT USE THESE AS ICAP EVIDENCE
==================================================

Do NOT determine the label from:

- punctuation
- question marks
- message length
- grammatical complexity
- technical vocabulary
- topic difficulty
- the existing llm_label
- chatId
- student's perceived intelligence
- assumed learning outcome
- assumed motivation
- assumed understanding


==================================================
QUICK REFERENCE
==================================================

Signal -> Label

"How do I...?" with no attempt
-> PASSIVE

"What is...?" with no attempt
-> PASSIVE

"Explain this..."
-> PASSIVE

"Fix/analyze/solve this..."
-> PASSIVE

"I don't understand..."
-> PASSIVE

"Here is my code, fix it."
-> PASSIVE

"Thanks/okay/yes."
-> PASSIVE

"I tried X and got Y."
-> ACTIVE

"I calculated X. Is it correct?"
-> ACTIVE

"I applied the formula and got X."
-> ACTIVE

"I changed X and observed Y."
-> ACTIVE

"So X is what the tutor meant?"
-> ACTIVE, if confirming information already provided

"I think X happens because Y."
-> CONSTRUCTIVE

"X will probably happen if Y."
-> CONSTRUCTIVE

"This result suggests X because Y."
-> CONSTRUCTIVE

"X is similar to Y because..."
-> CONSTRUCTIVE

"X should cause Y because..."
-> CONSTRUCTIVE

A question containing a NEW inference/reasoning
-> CONSTRUCTIVE

A question that only requests information
-> PASSIVE


==================================================
FINAL ANNOTATION TASK
==================================================

Previous AI/tutor message:
{context}

Student's message:
{student_turn}

Classify the student's message as exactly ONE of:

Passive
Active
Constructive


==================================================
OUTPUT FORMAT
==================================================

You MUST respond in EXACTLY these three lines:

LABEL: <Passive, Active, or Constructive>
REASON: <one concise sentence naming Q1 or Q2 and explaining the deciding evidence>
CONFIDENCE: <a number from 0.0 to 1.0>

Rules for REASON:
- exactly one sentence
- concise
- mention Q1 or Q2
- refer to observable evidence in the student's message
- do not invent information
- do not describe hidden motivation or learning outcome

The confidence represents your confidence in the annotation decision,
NOT the probability that the label is objectively correct.

Do not return multiple labels.
Do not use Interactive.
Do not use the existing llm_label to make the decision.
"""
````

````python
import time

def classify_with_retry(student_text, prior_context="", topic="", max_retries=10):

    for attempt in range(max_retries):

        try:

            result = classify_icap(
                student_text=student_text,
                prior_context=prior_context,
                topic=topic
            )

            return result

        except Exception as e:

            error_text = str(e)

            # Gemini rate limit
            if "429" in error_text or "RESOURCE_EXHAUSTED" in error_text:

                wait_time = min(60 * (attempt + 1), 300)

                print(
                    f"⚠️ Rate limit hit. "
                    f"Waiting {wait_time}s before retry..."
                )

                time.sleep(wait_time)

            else:
                print(f"❌ Error: {e}")

                # Wait briefly for temporary errors
                time.sleep(5)

    raise RuntimeError(
        "Maximum retries reached for this row."
    )
````

````python
ICAP_PROMPT_TEMPLATE = """
You are an expert annotator applying the ICAP framework to ONE student message in an AI-tutoring dialogue.

Your task is to classify the student's cognitive engagement as exactly ONE of:
PASSIVE, ACTIVE, CONSTRUCTIVE.

IMPORTANT:
Judge ONLY what the STUDENT contributes in the current message.
Do not classify based on the fact that the message is a question.
Do not treat pasted code, errors, assignment text, or copied material as the student's own cognitive work.
Use the prior tutor/AI message only to determine whether the student's idea is already present in the conversation.

CLASSIFICATION PROCEDURE

Q1 — Does the student's message contain their OWN cognitive work?

This includes:
- an attempted solution
- their own answer or result
- an answer they want checked
- their own interpretation
- a belief, inference, prediction, hypothesis, or explanation

If NO → PASSIVE.
If YES → go to Q2.

Q2 — Does the student generate NEW meaning beyond what was already provided?

This includes:
- their own reasoning or explanation
- a prediction or hypothesis
- a causal/mechanistic explanation
- a comparison with reasoning
- a connection between concepts
- an interpretation of a result
- an elaboration or self-explanation

If YES → CONSTRUCTIVE.
If NO → ACTIVE.

DEFINITIONS

PASSIVE:
The student mainly receives, requests, or delegates information without showing their own cognitive work.

Examples:
- "How do I reverse a list in Python?"
- "What is gradient descent?"
- "Explain this concept."
- "I don't understand polynomial features."
- "Help me with this."
- Pasting code/error/assignment and asking the AI to solve, fix, explain, analyze, or rewrite it.
- "Thanks", "ok", "idk".

ACTIVE:
The student shows some cognitive activity or performs an operation using available information, but does NOT generate new meaning or reasoning.

Examples:
- "I tried sorted(x) but it returns None. What did I do wrong?"
- "Is the answer B?"
- "I calculated 25, is that correct?"
- "So you're saying the loop runs n times?" when the tutor already stated this.
- Restating or applying information already provided without adding new reasoning.

CONSTRUCTIVE:
The student generates new meaning through their own reasoning, inference, prediction, explanation, interpretation, comparison, or connection.

Examples:
- "I think the loss plateaus because the learning rate is too small."
- "So a larger batch size might reduce noise but make each update slower?" when this idea was NOT already stated.
- "I think this happens because the model is overfitting to the training data."
- "The two methods are similar, but the second one should be faster because it avoids repeated computation."

IMPORTANT CONTEXT RULE:
A student's statement is CONSTRUCTIVE only when the reasoning, prediction, interpretation, comparison, or connection is generated by the student and is not already explicitly provided by the tutor/AI.

For example:
- Tutor: "A larger stride reduces the number of operations."
- Student: "So a larger stride will make training faster?"
→ ACTIVE, because the relevant relationship was already provided.

But:
- Tutor: "Stride controls how far the filter moves."
- Student: "So a larger stride might make training faster because the filter visits fewer positions?"
→ CONSTRUCTIVE, because the student generates the explanation.

TIE-BREAKING RULES:

1. Passive vs Active:
If the student's OWN attempt/work is not visible, choose PASSIVE.

2. Active vs Constructive:
If the student does not generate a new reason, prediction, interpretation, comparison, or connection, choose ACTIVE.

3. A question is NOT automatically Active or Constructive.
Classify according to the cognitive content of the question.

4. Pasted material is NOT student-generated work.
Judge the student's own words around the pasted material.

5. If the prior context is empty or insufficient:
Do not invent context. Judge the student's message using only the evidence available.

6. Code-related messages:
Do not automatically classify code as Active or Constructive.
If the student only asks the AI to explain/fix/analyze provided code → PASSIVE.
If the student reports their own attempted code/result → ACTIVE.
If the student explains why their code should work, predicts an outcome, or gives their own reasoning → CONSTRUCTIVE.

PRIOR AI/TUTOR MESSAGE:
{context}

STUDENT MESSAGE:
{student_turn}

Return ONLY valid JSON with exactly these fields:
{{
  "label": "Passive" | "Active" | "Constructive",
  "reason": "One concise sentence explaining the deciding evidence."
}}

The reason must explicitly refer to whether the student shows their own work (Q1) and/or generates new meaning (Q2).
"""

print("✅ ICAP_PROMPT_TEMPLATE loaded!")
````

````python
# ============================================================
# FEATURE ENGINEERING 1: LEXICAL FEATURES
# ============================================================

import re
import pandas as pd
import numpy as np

def extract_lexical_features(text):
    text = str(text)
    words = text.split()

    return {
        "word_count": len(words),
        "character_count": len(text),
        "sentence_count": len(re.findall(r"[.!?]+", text)),
        "question_count": text.count("?"),
        "exclamation_count": text.count("!"),
        "comma_count": text.count(","),
        "uppercase_count": sum(c.isupper() for c in text),
        "digit_count": sum(c.isdigit() for c in text),

        "first_person_count": len(re.findall(
            r"\b(i|i'm|i've|i'll|my|me|mine|we|our|us)\b",
            text.lower()
        )),

        "reasoning_word_count": len(re.findall(
            r"\b(because|therefore|so|thus|hence|since|why|reason|means|meaning|if|then)\b",
            text.lower()
        )),

        "uncertainty_word_count": len(re.findall(
            r"\b(maybe|perhaps|might|could|possibly|probably|think|believe|guess|suppose)\b",
            text.lower()
        ))
    }


lexical_features = labeled_df["prompt"].apply(
    extract_lexical_features
)

lexical_df = pd.DataFrame(
    lexical_features.tolist(),
    index=labeled_df.index
)

print("Lexical feature engineering completed.")
print("Number of lexical features:", lexical_df.shape[1])
print("\nFeatures:")
print(lexical_df.columns.tolist())

display(lexical_df.head())
````
